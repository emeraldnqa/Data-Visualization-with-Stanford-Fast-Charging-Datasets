Using Mari's code, I looking at the batteries in channel 29

```python
import json 
import pprint
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Read the json file
f = open('2018-08-28_oed_0_CH29_structure.json')
chanel_29_data = json.load(f)
```

```python
#Print the highest level in the dictionary
for k in chanel_29_data.keys():
     print(k)
```

```python
#See what is inside the "summary" key of the dictionary
for s in chanel_29_data['summary']:
    print (s)
```

```python
#Look specifically into the data of the discharge_capacity inside the summary key of dictionary 
for x in chanel_29_data['summary']['discharge_capacity']:
        print(x)
```

```python
# Concert the summary part in the json file into a dataframe (I copied these codes from Beren's)
summary_828ch029 = chanel_29_data["summary"]
summary_828ch029_df = pd.DataFrame.from_dict(summary_828ch029)
summary_828ch029_df.head()
```

```python
summary_828ch029_df.describe()
```

Here I want to plot graph visualizing the relationship between the charge capacity vs charge duration along with the relationship between the discharge capacity vs charge duration in the graph

```python
ch29_charge_duration = summary_828ch029['charge_duration']
ch29_charge_capacity = summary_828ch029['charge_capacity']
ch29_discharge_capacity = summary_828ch029['discharge_capacity']
fig = plt.figure()
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.plot(ch29_charge_capacity,ch29_charge_duration)
plt.title("Charge Capacity vs Charge Duration")
plt.xlabel("Charge Capacity")
plt.ylabel("Charge Duration")

plt.subplot(2,2,2)
plt.plot(ch29_discharge_capacity,ch29_charge_duration)
plt.title("Discharge Capacity vs Charge Duration")
plt.xlabel("Discharge Capacity")
plt.ylabel("Charge Duration")
plt.show()
```

- I'm not sure if there's a direct relations between discharge capacity and charge duration, both variable when graphed out, is linear.

- Now we may want to see a relationship between the charge duration and voltage, since the voltage is controlled during charging cycle. However, since I don't see voltage summary data in here, so I opened the raw data key to see if they have voltage

```python
#Open raw_data key in the dictionary:
for r in chanel_29_data['raw_data']:
    print (r)
```

They do have voltage, charge capacity and current in here. So now I will convert the raw data in json into a data frame

```python
# Concert the raw_data part in the json file into a dataframe (I copied these codes from Beren's)
raw_data_828ch029 = chanel_29_data["raw_data"]
raw_data_828ch029_df = pd.DataFrame.from_dict(raw_data_828ch029)
raw_data_828ch029_df.head()
```

```python
raw_data_828ch029_df.describe()
```

- All the properties of the batteries that I'm trying to visualize is based on Peter's paper. Mostly on current, voltage, discharge capacity, charge capacity and step time.

- The paper also talks about battery cycle life, but the data set doesn't talk about it here
- Most of the relationship graph here are lines so I'm wondering would they look different with scatter plot

```python
ch29_raw_data_step_time = raw_data_828ch029['step_time']
ch29_raw_data_charge_capacity = raw_data_828ch029['charge_capacity']
ch29_raw_data_discharge_capacity = raw_data_828ch029['discharge_capacity']
fig = plt.figure()
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.plot(ch29_raw_data_step_time, ch29_raw_data_charge_capacity,)
plt.title("Charge Capacity vs Step time")
plt.xlabel("Step time")
plt.ylabel("Charge Capacity")

plt.subplot(2,2,2)
plt.plot(ch29_raw_data_step_time, ch29_raw_data_discharge_capacity)
plt.title("Discharge Capacity vs Step time")
plt.xlabel("Step time")
plt.ylabel("Discharge Capacity")
plt.show()
```

```python
ch29_raw_data_step_time = raw_data_828ch029['step_time']
ch29_raw_data_voltage = raw_data_828ch029['voltage']
ch29_raw_data_current = raw_data_828ch029['current']
fig = plt.figure()
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.plot(ch29_raw_data_step_time, ch29_raw_data_voltage)
plt.title("Voltage vs Step time")
plt.xlabel("Step time")
plt.ylabel("Voltage")

plt.subplot(2,2,2)
plt.plot(ch29_raw_data_step_time, ch29_raw_data_current)
plt.title("Current vs Step time")
plt.xlabel("Step time")
plt.ylabel("Current")
plt.show()
```
