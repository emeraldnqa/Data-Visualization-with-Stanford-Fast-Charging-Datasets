```python
import json 
import pprint
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Read the json file
# Read the json file
f = open('2018-08-28_oed_0_CH44_structure.json')
chanel_44_data = json.load(f)
```

```python
#Print the highest level in the dictionary
for k in chanel_44_data.keys():
     print(k)
```

Here I'll look through the Channel 44 the same as the information that I was looking through Channel 29

```python
# Concert the summary part in the json file into a dataframe (I copied these codes from Beren's)
summary_828ch044 = chanel_44_data["summary"]
summary_828ch044_df = pd.DataFrame.from_dict(summary_828ch044)
summary_828ch044_df.head()
```

```python
summary_828ch044_df.describe()
```

```python
ch44_charge_duration = summary_828ch044['charge_duration']
ch44_charge_capacity = summary_828ch044['charge_capacity']
ch44_discharge_capacity = summary_828ch044['discharge_capacity']
fig = plt.figure()
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.plot(ch44_charge_capacity,ch44_charge_duration)
plt.title("Charge Capacity vs Charge Duration")
plt.xlabel("Charge Capacity")
plt.ylabel("Charge Duration")

plt.subplot(2,2,2)
plt.plot(ch44_discharge_capacity,ch44_charge_duration)
plt.title("Discharge Capacity vs Charge Duration")
plt.xlabel("Discharge Capacity")
plt.ylabel("Charge Duration")
plt.show()
```
